```python
# Q 7 
arr = np.arange(1, 13).reshape(3, 4)

# Original array
print(arr)

# a) Shape
print("Shape:", arr.shape)

# b) Transpose
print("Transpose:")
print(arr.T)

# c) Reshape into 5x5
try:
    print(arr.reshape(5, 5))
except ValueError as e:
    print("Error:", e)
```

```python
# Q 8

arr = np.array([1, 2, 3, 4])

# a) Integer to float
float_arr = arr.astype(float)
print(float_arr)
print(float_arr.dtype)

# b) Float to string
str_arr = float_arr.astype(str)
print(str_arr)
print(str_arr.dtype)

# c) Original array ka data type
print(arr.dtype)
```

```python
# Q 9

arr = np.array([[1, 2], [3, 4]])

# a) Har row par loop
for row in arr:
    print(row)

# b) Har element par nditer()
for value in np.nditer(arr):
    print(value)
```

```python
# Q 10

arr = np.array([[1, 2], [3, 4]])

# a) Row-wise sum
for row in arr:
    print("Row sum:", np.sum(row))

# b) Har element ka square
for value in np.nditer(arr):
    print(value ** 2)
```

```python
#Q 11

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)

print("Subtraction:", a - b)

print("Multiplication:", a * b)

print("Division:", a / b)
```

```python
#Q12

a = np.array([1, 4, 9, 16])

# a) Square root
print("Square root:", np.sqrt(a))

# b) Power 2
print("Power 2:", a ** 2)

# c) Natural logarithm
print("Logarithm:", np.log(a))
```

```python
#Q 13

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# a) Element-wise multiplication
print("Element-wise:")
print(A * B)

# b) Matrix multiplication
print("Matrix multiplication:")
print(A @ B)

# Alternative
print(np.dot(A, B))
```

```python
# Q14

arr = np.random.randint(0, 21, size=10)
print("Original:", arr)

# a) Values greater than 10
print("Greater than 10:", arr[arr > 10])

# b) Count even values
print("Even count:", np.sum(arr % 2 == 0))

# c) Replace values less than 5 with 0
arr[arr < 5] = 0
print("Updated:", arr)
```
