```python
def analyze_string(text):
   characters = len(text)
   vowels = 0
   spaces = 0

   for ch in text:  
       if ch.lower() in "aeiou":  
        vowels += 1  
       if ch == " ":  
        spaces += 1  

   return characters, vowels, spaces

result = analyze_string("Python Programming")

print("Characters:", result[0])
print("Vowels:", result[1])
print("Spaces:", result[2])
```

```python
balance = 10000

def deposit(amount):
 global balance
 balance = balance + amount
 print("Deposited:", amount)

def withdraw(amount):
    global balance

    if amount <= balance:  
       balance = balance - amount  
       print("Withdrawn:", amount)  
    else:  
         print("Insufficient Balance")


def check_balance():
    print("Balance:", balance)

deposit(2000)
check_balance()

withdraw(5000)
check_balance()

withdraw(10000)
check_balance()
```

```python
def calculate_result(name, *marks):
   total = sum(marks)
   average = total / len(marks)
   highest = max(marks)
   lowest = min(marks)
 
   if all(mark >= 40 for mark in marks):  
    result = "Pass"  
   else:  
    result = "Fail"  

   return name, total, average, highest, lowest, result

result = calculate_result("Rahul", 80, 70, 90, 55, 65)

print("Name:", result[0])
print("Total:", result[1])
print("Average:", result[2])
print("Highest:", result[3])
print("Lowest:", result[4])
print("Result:", result[5])
```

```python
def analyze_numbers(*numbers):
 total = sum(numbers)
 average = total / len(numbers)
 maximum = max(numbers)
 minimum = min(numbers)

 even = 0  
 odd = 0  
 positive = 0  
 negative = 0  

 for number in numbers:  

    if number % 2 == 0:  
        even += 1  
    else:  
        odd += 1  

    if number > 0:  
        positive += 1  
    elif number < 0:  
        negative += 1  

 return total, average, maximum, minimum, even, odd, positive, negative

result = analyze_numbers(10, -5, 20, -7, 3, 8)

print("Sum:", result[0])
print("Average:", result[1])
print("Maximum:", result[2])
print("Minimum:", result[3])
print("Even:", result[4])
print("Odd:", result[5])
print("Positive:", result[6])
print("Negative:", result[7])
```
