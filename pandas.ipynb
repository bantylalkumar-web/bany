```python
import pandas as pd

data = {
    'Name': ['Amit', 'Rohit', 'Neha', 'Priya', 'Rahul'],
    'Age': [21, 23, 25, 22, 24],
    'Salary': [25000, 35000, 40000, 28000, 45000]
}

df = pd.DataFrame(data, index=['a', 'b', 'c', 'd', 'e'])

# Display DataFrame
print(df)

```

```python
# 1. Select only one row
print(df.iloc[1])


```

```python
# 2. Select multiple rows
print(df.loc[['b', 'd']])

```

```python
# 3. Select a specific row and column
print(df.loc['b', 'Name'])

```

```python
# 4. Select one row with multiple columns
print(df.loc['b', ['Name', 'Salary']])

```

```python
# 5. Select all rows and one column
print(df.loc[:, 'Name'])

```

```python
# 6. Select all rows with multiple columns
print(df.loc[:, ['Name', 'Age']])

```

```python
# 7. Select rows from index 'b' to 'd'
print(df.loc['b':'d', ['Name', 'Salary']])

```

```python
 #8. Select rows where Age > 22
print(df.loc[df['Age'] > 22])

```

```python
# 9. Select rows where Age > 22 and Salary > 35000
print(df.loc[(df['Age'] > 22) & (df['Salary'] > 35000)])

```

```python
# 10. Select Name and Salary where Salary > 30000
print(df.loc[df['Salary'] > 30000, ['Name', 'Salary']])

```

```python
# 11. Update salary for index 'a'
df.loc['a', 'Salary'] = 28000

```
